```python
import os
import pandas as pd
import json
import hashlib
```

```python
def get_csv_files(filepath):
    csvs = []
    for subdir, dirs, files in os.walk(filepath):
        files.sort()
        for file in files:
            filepath = subdir + os.sep + file
            if filepath.endswith(".csv"):
                csvs.append(filepath)
    return csvs

def get_urls(suburb_csv):
    address_for_sales = get_csv_files(suburb_csv)
    print(f'Number of suburbs with sale properties: {len(address_for_sales)}')
    SaleURLs = []
    for suburb_csv in address_for_sales:
        try:
            df = pd.read_csv(suburb_csv)
            SaleURLs.extend(df['SaleURL'].tolist())
        except:
            continue
    print(f'Number of sale properties: {len(SaleURLs)}')
    return SaleURLs
```

# Make batches

```python
import json

# Load the list of URLs from the input file
input_file = 'urls_sale_QLD_existing_run18.json'
with open(input_file, 'r') as f:
    urls = json.load(f)

# Split into 8 nearly equal parts
batch_size = len(urls) // 8
for i in range(8):
    start = i * batch_size
    end = (i + 1) * batch_size if i < 7 else len(urls)  # Last batch takes the remainder
    batch = urls[start:end]
    
    output_file = f'urls_sale_QLD_existing_run18_batch{i+1}.json'
    with open(output_file, 'w') as f:
        json.dump(batch, f, indent=2)

print(f"Split {len(urls)} URLs into 8 batches.")
```

```python
import re
from pydantic import EmailStr, ValidationError

def validate_email(email):
        """
        Validates the email address by ensuring the domain (after the '@') contains a period.
        
        Parameters:
            email (str): The email address to validate.
        
        Returns:
            str or None: Returns the original email if valid, otherwise returns None.
        """
        # Ensure the input is a string and contains an '@'
        if not isinstance(email, str) or '@' not in email:
            return None

        try:
            # Split the email into local and domain parts
            local_part, domain_part = email.split('@', 1)
        except ValueError:
            # If splitting fails, the email is not valid
            return None

        # Check if the domain part contains at least one period
        if '.' not in domain_part:
            return None

        # Email passes the basic validation check
        return email

def clean_email(email: str) -> str | None:
    # Step 1: If full name + email, keep only the email part
    if ' ' in email:
        email = email.split()[-1]

    # Step 2: Remove trailing noise like '-1' after the domain
    match = re.match(r'^(.+@[^@\s]+?)(-[0-9]+)?$', email)
    if match:
        email = match.group(1)

    # Step 2.5: Remove '.' immediately before '@'
    email = re.sub(r'\.(?=@)', '', email)

    # Step 3: Validate using Pydantic
    if not validate_email(email):
        email = None

    return email
```

```python
import pandas as pd

def count_properties_by_suburb(df):
    # Group by state and suburb, then count
    counts = df.groupby(['state', 'suburb']).size().reset_index(name='#properties')
    
    # Optional: sort by number of properties descending
    counts = counts.sort_values(by='#properties', ascending=False).reset_index(drop=True)
    
    return counts
```
